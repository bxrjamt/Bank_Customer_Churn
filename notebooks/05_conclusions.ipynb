```python
import pandas as pd

# Leer el archivo .txt
df = pd.read_csv(r"C:\Users\Borja\Desktop\ADP2.txt", sep=",", encoding="utf-8")

# Reemplazar ' | ' por salto de línea en Opciones
df["Opciones"] = df["Opciones"].str.replace(" \| ", "\n", regex=True)

# Función para obtener el texto de la respuesta correcta incluyendo la letra
def obtener_texto_respuesta(row):
    letra = row["Respuesta correcta"].strip()  # "a", "b" o "c"
    opciones = row["Opciones"].split("\n")     # Lista de opciones
    for opcion in opciones:
        if opcion.lower().startswith(letra.lower() + "."):
            return opcion  # Devolver la opción completa con la letra
    return ""

# Aplicar la función
df["Respuesta correcta"] = df.apply(obtener_texto_respuesta, axis=1)
df
# Guardar como CSV
#df.to_csv(r"C:\Users\Borja\Desktop\preguntas_con_saltos2.csv", index=False, encoding="utf-8-sig")
```
